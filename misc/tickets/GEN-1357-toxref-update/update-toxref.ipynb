### Get Data from API

```python
from requests import get, post
```

```python
dtxsid = "DTXSID7034676"
effect_url = f"https://ctx-api-stg.ccte.epa.gov/hazard/toxref/data/search/by-dtxsid/{dtxsid}"
resp = get(effect_url)
effect_api = resp.json()
JSON(effect_api)
```

```python
observation_url = f"https://ctx-api-stg.ccte.epa.gov/hazard/toxref/observations/search/by-dtxsid/{dtxsid}"
obs_resp = get(observation_url)
obs_api = obs_resp.json()[0]
JSON(obs_api)
```

```python
all_api_keys = set()
for doc in resp.json():
    all_api_keys.update(doc.keys())

all_db_keys = set()
for doc in DB.toxref_effects.find({"dsstox_sid": dtxsid},{"_id": 0}):
    all_db_keys.update(doc.keys())
```
