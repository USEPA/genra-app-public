### Create a collection that contains DTXSIDs which have Aquatic EcoTox data

```python
import sys
PATH_TO_TOP = "/genra"
if PATH_TO_TOP not in sys.path:
    sys.path.append(PATH_TO_TOP)
import os
os.chdir(PATH_TO_TOP)
from IPython.display import JSON
from genraweb.resources import DB
import pandas as pd
```

```python
# load data from excel
ecotox_df = pd.read_excel("attic/ecotox-chems-dec-2024-COUNTS_STD.xlsx")
ecotox_df.head()
# TODO: anything we should do about entries without a dtxsid?
```

```python
# get sid where aquatic data exists
etaq_ids = ecotox_df.DTXSID[ecotox_df.RECS_STD_AQUATIC > 0].dropna()
etaq_ids.values
```

```python
# format list for db
docs = []
for sid in etaq_ids:
    docs.append({"dsstox_sid": sid})
```

```python
# insert data into collection
DB.tox_etaq.delete_many({})
DB.tox_etaq.insert_many(docs)
```
