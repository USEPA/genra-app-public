# Mean similarity by FP type

To validate the observation the similarities for HTPP FP (2022-10) run lower then other FPs and require a lower default s0.

Calculates mean similarity for 15 nearest neighbors by FP type.

```python
# setup
import sys
import os
if "/genra" not in sys.path:
    sys.path.append("/genra")
from IPython.display import JSON
API_URL = "http://genra_api:5000/genra-api/api/genra/"
```

```python
import numpy as np
import matplotlib.pyplot as pl
import pandas as pd
import math
from genraweb.resources import DB
```

```python
JSON(DB.fp_info.find_one({"dsstox_cid":"DTXCID30182"}, {"_id":0}))
```

```python
from genraweb.lib.fp.fpclass import FPGen
bios = [k for k, v in FPGen.FPClass.items() if v.similarity_tag in 'bt']
q = [{k: {"$ne": None}} for k in bios]
q = {"$or": q}
count = DB.fp_info.count_documents(q)
count
```

```python
# run this for "tox_txrf", "bio_txct", and "no_filter"
filter_ = "tox_txrf"
from collections import defaultdict
obs = defaultdict(list)
for rec in DB.fp_info.find(q):
    for fp in FPGen.FPClass:
        if fp in rec:
            if filter_ in rec[fp]:  # no_filter not always present
                obs[fp].extend(rec[fp][filter_]['similarities'][:15])
```

```python
print(15*count)  # max possible
sims = []
for fp in obs:
    sims.append((fp, len(obs[fp]), round(sum(obs[fp]) / len(obs[fp]), 3)))
sims.sort(key=lambda x: x[2], reverse=True)
sims
```
